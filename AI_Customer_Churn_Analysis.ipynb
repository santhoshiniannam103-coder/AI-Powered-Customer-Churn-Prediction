```python
# X = Features
# These are the customer details that the AI model
# will use to predict whether a customer will churn.
# We remove:
# "Churn" → Original Yes/No answer
# "Churn_Encoded" → Numerical version of the answer
# We remove them because they are the target, not input features.
X = df.drop(["Churn", "Churn_Encoded"], axis=1)
```

```python
# y = Target
# This is the value that our AI model needs to predict.
# 0 = Customer did NOT churn
# 1 = Customer churned
y = df["Churn_Encoded"]
```

```python
# Check the number of rows and columns in X and y
print("Features (X) shape:", X.shape)
print("Target (y) shape:", y.shape)
```

```python
# Import the function used to split our dataset
from sklearn.model_selection import train_test_split
```

```python
# Split the data into:
# 80% → Training data
# 20% → Testing data
# stratify=y keeps the same proportion of churn and non-churn
# customers in both training and testing datasets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
```

```python
# Check the size of each dataset
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# Check churn distribution in training data

print("Training Churn Distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting Churn Distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
# Encoding Categorcal Features
# Import OneHotEncoder
# It converts text categories into numerical columns

from sklearn.preprocessing import OneHotEncoder
```

```python
# Identify categorical and numerical columns
# Find columns containing text/categories
categorical_cols = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

# Find columns containing numerical values
numerical_cols = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("Categorical columns:")
print(categorical_cols)

print("\nNumerical columns:")
print(numerical_cols)
```

```python
# Creating One-Hot Encoder
# OneHotEncoder converts categorical/text values
# into numerical 0 and 1 columns.
#
# handle_unknown="ignore" prevents errors if the
# test data contains a category not seen during training.

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)
```

```python
# Fit the encoder using only the training data
# and transform the training categorical columns.

X_train_encoded = encoder.fit_transform(
    X_train[categorical_cols]
)
```

```python
# Transform the test categorical columns
# using the encoder already learned from training data.

X_test_encoded = encoder.transform(
    X_test[categorical_cols]
)
```

```python
print("Encoded training shape:", X_train_encoded.shape)
print("Encoded testing shape:", X_test_encoded.shape)
```

```python
X = df.drop(
    ["customerID", "Churn", "Churn_Encoded"],
    axis=1
)

y = df["Churn_Encoded"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Combine numerical and encoded categorical features

X_train_numeric = X_train[numerical_cols].values
X_test_numeric = X_test[numerical_cols].values

X_train_final = np.hstack([
    X_train_numeric,
    X_train_encoded
])

X_test_final = np.hstack([
    X_test_numeric,
    X_test_encoded
])

print("Final training features:", X_train_final.shape)
print("Final testing features:", X_test_final.shape)
```

```python
# Importing StandardScalar
# StandardScaler standardizes numerical features
# so they have a similar scale.

from sklearn.preprocessing import StandardScaler
```

```python
# Creating the scaler
scaler = StandardScaler()
```

```python
# Fit the scaler using ONLY the training numerical data
# and transform the training data.

X_train_numeric_scaled = scaler.fit_transform(
    X_train[numerical_cols]
)
```

```python
# Transforming the test numerical data
# using the scaler learned from training data.

X_test_numeric_scaled = scaler.transform(
    X_test[numerical_cols]
)
```

```python
# Combining Scaled numerical and Encoding categorial

X_train_final = np.hstack([
    X_train_numeric_scaled,
    X_train_encoded
])

X_test_final = np.hstack([
    X_test_numeric_scaled,
    X_test_encoded
])
print("Final training features:", X_train_final.shape)
print("Final testing features:", X_test_final.shape)
```

```python
# Logistic Regression
# This model predicts whether a customer will churn or not churn.

from sklearn.linear_model import LogisticRegression
```

```python
# Creating the Logistic Regression model

logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)
```

```python
# Training the Logistic Regression model
# X_train_final = customer features
# y_train       = actual churn answers

logistic_model.fit(
    X_train_final,
    y_train
)

print("Logistic Regression model trained successfully!")
```

```python
# Making predictions on unseen test customers

y_pred = logistic_model.predict(X_test_final)

print("First 10 predictions:")
print(y_pred[:10])
```

```python
# Model Evaluation
# Import accuracy_score
from sklearn.metrics import accuracy_score

# Comparing predicted values with actual values
accuracy = accuracy_score(y_test, y_pred)

# Displaying accuracy as a percentage
print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
# Model Evaluation - Classification Report
from sklearn.metrics import classification_report

# Displaying Precision, Recall, F1-score and Support
print(classification_report(y_test, y_pred))
```

```python
# Random Forest Model
# Random Forest uses multiple decision trees
# to make a final prediction.

from sklearn.ensemble import RandomForestClassifier
```

```python
# Creating the Random Forest model
random_forest = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)
```

```python
# Training Random Forest
random_forest.fit(
    X_train_final,
    y_train
)

print("Random Forest model trained successfully!")
```

```python
# predictions on test data
y_pred_rf = random_forest.predict(X_test_final)

print("First 10 Random Forest predictions:")
print(y_pred_rf[:10])
```

```python
# Evaluating Random Forest
from sklearn.metrics import accuracy_score, classification_report

# Calculating accuracy
accuracy_rf = accuracy_score(y_test, y_pred_rf)

print(f"Random Forest Accuracy: {accuracy_rf * 100:.2f}%")

# Displaying precision, recall and F1-score
print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
# XGBoost Model
# XGBoost is a powerful algorithm for classification problems such as customer churn.

from xgboost import XGBClassifier
```

```python
# Creating the XGBoost model

xgb_model = XGBClassifier(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric="logloss"
)
```

```python
# Training the XGBoost model using the training data

xgb_model.fit(
    X_train_final,
    y_train
)

print("XGBoost model trained successfully!")
```

```python
# Predict churn for the unseen test customers

y_pred_xgb = xgb_model.predict(X_test_final)

print("First 10 XGBoost predictions:")
print(y_pred_xgb[:10])
```

```python
# Evaluating XGBoost
from sklearn.metrics import accuracy_score, classification_report

# Calculating accuracy
accuracy_xgb = accuracy_score(y_test, y_pred_xgb)

print(f"XGBoost Accuracy: {accuracy_xgb * 100:.2f}%")

# Displaying Precision, Recall and F1-score
print("\nClassification Report:")
print(classification_report(y_test, y_pred_xgb))
```

```python
# Tuned XGBoost model
# Creating a new XGBoost model with adjusted parameters.
# We are keeping the original XGBoost model unchanged
# so we can compare the results later.

xgb_tuned = XGBClassifier(
    n_estimators=300,
    max_depth=3,
    learning_rate=0.03,
    subsample=0.9,
    colsample_bytree=0.9,
    random_state=42,
    eval_metric="logloss"
)

print("Tuned XGBoost model created successfully!")
```

```python
# Training Tuned XGBoost
# Training the tuned model using the same training data

xgb_tuned.fit(
    X_train_final,
    y_train
)

print("Tuned XGBoost model trained successfully!")
```

```python
# Predicting churn for the test customers

y_pred_xgb_tuned = xgb_tuned.predict(X_test_final)

print("First 10 tuned XGBoost predictions:")
print(y_pred_xgb_tuned[:10])
```

```python
# Evaluating Tuned XGBoost
from sklearn.metrics import accuracy_score, classification_report

# Calculating accuracy
accuracy_xgb_tuned = accuracy_score(
    y_test,
    y_pred_xgb_tuned
)

print(
    f"Tuned XGBoost Accuracy: "
    f"{accuracy_xgb_tuned * 100:.2f}%"
)

# Displaying Precision, Recall and F1-score
print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_xgb_tuned
    )
)
```

```python
# Churn Probability
# Get the probability of each customer churning.
# [:, 1] selects the probability of class 1.
# Class 1 means "Churn".

y_prob_xgb = xgb_tuned.predict_proba(X_test_final)[:, 1]

# Displaying the first 10 churn probabilities

print("First 10 Churn Probabilities:")
print(y_prob_xgb[:10])
```

```python
# Creating Customer Risk Level
import numpy as np

# Creating risk categories based on churn probability

risk_level = np.where(
    y_prob_xgb < 0.30,
    "Low Risk",
    np.where(
        y_prob_xgb < 0.70,
        "Medium Risk",
        "High Risk"
    )
)

# Displaying the first 10 risk levels

print("First 10 Customer Risk Levels:")
print(risk_level[:10])
```

```python
# Counting Customers By Risk Level
risk_counts = pd.Series(risk_level).value_counts()

print("Customer Risk Distribution:")
print(risk_counts)
```

```python
# Creating Customer Risk Analysis Table
# Getting the original customer IDs from the test dataset
customer_ids = df.loc[X_test.index, "customerID"]

# Creating a new DataFrame containing prediction results
risk_analysis = pd.DataFrame({
    "customerID": customer_ids.values,
    "Actual_Churn": y_test.values,
    "Predicted_Churn": y_pred_xgb,
    "Churn_Probability": y_prob_xgb,
    "Risk_Level": risk_level
})

# Displaying the first 10 customers
risk_analysis.head(10)
```

```python
# Adding important Customer Information
# Selecting useful business-related columns
customer_details = df.loc[
    X_test.index,
    [
        "gender",
        "SeniorCitizen",
        "Partner",
        "Dependents",
        "tenure",
        "Contract",
        "InternetService",
        "MonthlyCharges",
        "TotalCharges",
        "TotalServices"
    ]
].reset_index(drop=True)
risk_analysis = risk_analysis.reset_index(drop=True)

# Combining customer details with prediction results
risk_analysis = pd.concat(
    [risk_analysis, customer_details],
    axis=1
)

# Displaying the first 10 rows
risk_analysis.head(10)
```

```python
# Finding High Risk Customers
# Selecting customers with High Risk of Churn
high_risk_customers = risk_analysis[
    risk_analysis["Risk_Level"] == "High Risk"
].sort_values(
    by="Churn_Probability",
    ascending=False
)

# Displaying the top 10 highest-risk customers

high_risk_customers.head(10)
```
